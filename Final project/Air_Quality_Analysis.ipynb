```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
# 1. Dataset Creation

df = pd.read_csv('air_quality_data.csv')


print("=== Air Quality Summary ===")
print(df.head())

```

```python
# 2. Visualization

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot 1: City wise AQI Level

sns.barplot(ax=axes[0], data=df, x='City', y='AQI', palette='magma')
axes[0].set_title('Air Quality Index (AQI) by City', fontsize=12, fontweight='bold')
axes[0].axhline(100, color='red', linestyle='--', label='Unhealthy Threshold (>100)')
axes[0].legend()

# Plot 2: PM2.5 vs PM10 Comparison

df.plot(x='City', y=['PM2.5', 'PM10'], kind='line', ax=axes[1], colormap='viridis')
axes[1].set_title('PM2.5 vs PM10 Levels across Cities', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Concentration (µg/m³)')

plt.tight_layout()
plt.show()
```
